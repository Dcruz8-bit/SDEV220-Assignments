```python
from datetime import date

today = date.today()

with open("today.txt", "w") as file:
    file.write(str(today))

print("Wrote to today.txt:", today)
```

```python
import multiprocessing
import time
import random
from datetime import datetime

def worker():
    delay = random.random()
    time.sleep(delay)
    print(f"{multiprocessing.current_process().name} | time: {datetime.now().time()} | delay: {delay:.2f}s")

if __name__ == "__main__":
    processes = []
    for i in range(3):
        p = multiprocessing.Process(target=worker, name=f"Worker-{i+1}")
        processes.append(p)
        p.start()

    for p in processes:
        p.join()
```
